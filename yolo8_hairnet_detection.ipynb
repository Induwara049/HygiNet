```python
from ultralytics import YOLO
import os
```

```python
model = YOLO("best_hairnet.pt")
```

```python
test_img_dir = "dataset/hairnet/test/images"
```

```python
# Run inference on test folder
results = model.predict(source=test_img_dir, save=True, conf=0.25)
```
